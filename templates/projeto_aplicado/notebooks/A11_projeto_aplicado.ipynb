## 0. Configuração e reprodutibilidade

```python
from pathlib import Path

from projeto_aplicado.config import load_config
from projeto_aplicado.manifest import build_manifest, save_manifest

cfg = load_config('../config/projeto.yml')
print('Projeto:', cfg['projeto']['titulo'])
print('Semente:', cfg['reproducao']['seed'])
```

## 1. Carregamento dos dados

```python
import geopandas as gpd

gdf = gpd.read_file(
    cfg['data']['path'],
    layer=cfg['data'].get('layer'),
).to_crs(cfg['data']['analysis_crs'])

print(f'{len(gdf)} unidades territoriais, CRS: {gdf.crs}')
gdf.head()
```

## 2. Análise exploratória

<!-- Descreva o que os mapas e estatísticas mostram -->

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 6))
gdf.plot(
    column=cfg['data']['value_column'],
    ax=ax,
    legend=True,
    cmap='YlOrRd',
)
ax.set_title(cfg['data']['value_column'])
ax.set_axis_off()
plt.tight_layout()
plt.savefig('../outputs/eda/mapa_indicador.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Estatísticas descritivas
gdf[[cfg['data']['value_column']]].describe()
```
